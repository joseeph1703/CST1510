*Walkthrough: `02 - Parameters and Arguments`*

### D5. Arguments are matched by position

1. Define `within_limit(value, limit)` that returns `value <= limit`.
2. Print `within_limit(87, 100)`, then `within_limit(100, 87)`.
3. Add a comment explaining why the two results are different.

**Expected output:** `True`, then `False`

````python
# D5
# D5
# The arguments are matched by position.
# In the first call, 87 <= 100 is True.
# In the second call, 100 <= 87 is False because the values are reversed.

def within_limit(value, limit):
    return value <= limit

print(within_limit(87, 100))
print(within_limit(100, 87))
````

### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:
   - `"OVER LIMIT"` if `percent` is 100 or more
   - `"WARNING"` if `percent` is `warning_at` or more
   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python
# D6
def status_of(percent, warning_at=90):
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"

print(status_of(95))       # uses default warning_at = 90
print(status_of(85))       # still uses default
print(status_of(85, 80))   # overrides warning_at
````

### D11. Get the value out with `return`

This code should print `120.0`, but it prints `100`. Fix it using `return`, without changing the
`price = price * 1.2` line.

**Expected output after your fix:** `120.0`

````python
# D11
# Fix: the function must return the new price, otherwise the change is lost.

total = 100

def add_tax(price):
    price = price * 1.2
    return price   # send the updated value back

total = add_tax(total)
print(total)
````

### D12. Returning more than one value

1. Define `check_record(value, limit)` that works out `difference = value - limit` and
   `percent = (value / limit) * 100`, and returns **both**: `return difference, percent`
2. Call it with `check_record(87.0, 100.0)` and store the results in two variables.
3. Print them.

**Expected output:**
```
Difference: -13.0
Percent: 87.0
```

````python
# D12
# D12

def check_record(value, limit):
    difference = value - limit
    percent = (value / limit) * 100
    return difference, percent

diff, pct = check_record(87.0, 100.0)

print("Difference:", diff)
print("Percent:", pct)
````

*These combine several ideas from this week. There are no step-by-step instructions.*

### C1. Wire two functions together

Using `percent_of` from D3 and `status_of` from D6 **together**, print a status for each of these
three pairs. Do not write the percentage or status logic a second time.

| value | limit | Expected output |
|---|---|---|
| 87 | 100 | `OK` |
| 120 | 100 | `OVER LIMIT` |
| 45 | 100 | `OK` |

````python
# C1

def percent_of(value, limit):
    return (value / limit) * 100

def status_of(percent):
    if percent > 100:
        return "OVER LIMIT"
    else:
        return "OK"

values = [(87, 100), (120, 100), (45, 100)]

for value, limit in values:
    percent = percent_of(value, limit)
    status = status_of(percent)
    print(status)


````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python

# C2

def count_over(*values, limit=100):
    """Return how many values are greater than the limit."""
    return sum(1 for v in values if v > limit)

print(count_over(87, 120, 45, 101))                 # 2
print(count_over(87, 120, 45, 101, limit=50))       # 3
print(count_over())                                 # 0
````
